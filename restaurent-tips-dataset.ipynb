# Cat plot with Point plot

**seaborn abbrevation is sns** 

**kind fun use for draw diff graph**

```python
sns.catplot(x='tip',y='sex',data=tips,hue='sex',palette='BuPu',kind='point')
plt.show()
```

# Cat plot With Box plot

```python
sns.catplot(x='smoker',y='tip',data=tips,hue='sex',palette='BuPu',kind='box')
plt.show()
```

# Cat plot With Boxen Plot

```python
sns.catplot(x='day',y='size',data=tips,hue='sex',palette='BuPu',kind='boxen')
plt.show()
```

# Cat Plot With Strip plot

```python
sns.catplot(x='time',y='size',data=tips,hue='sex',palette='BuPu',kind='strip')
plt.show()
```

# Cat PLot With Swarm plot

```python
sns.catplot(x='time',y='day',data=tips,hue='sex',palette='BuPu',kind='swarm')
plt.show()
```

# Cat plot With Count Plot

```python
sns.catplot(x='time',data=tips,hue='sex',palette='BuPu',kind='count')
plt.show()
```

# Cat plot With Violine plot

```python
sns.catplot(x='smoker',y='size',data=tips,hue='sex',palette='BuPu',kind='violin')
plt.show()
```

# cat plot with bar plot

**palette fun use for diff color combination**

```python
sns.catplot(x='time',y='tip',data=tips,hue='sex',palette='BuPu',kind='bar')
plt.show()
```
